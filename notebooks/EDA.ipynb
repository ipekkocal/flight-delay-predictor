```python
import pandas as pd

# Load dataset
df = pd.read_csv("../data/flights.csv")

# Preview the first few rows
df.head()
```

```python
# Check data types, missing values, memory usage
df.info()

# Check missing values more explicitly
df.isnull().sum()
```

```python
# Get summary statistics for numerical columns
df.describe()
```

```python
import matplotlib.pyplot as plt

# Convert Month to datetime
df['Month'] = pd.to_datetime(df['Month'])

# Set as index for time-series plotting
df.set_index('Month', inplace=True)

# Line plot for passengers over time
df['Passengers'].plot(figsize=(12, 5), title="Monthly Airline Passengers", ylabel="Passengers", grid=True)
plt.show()
```
